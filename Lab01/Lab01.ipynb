Task 7

```python
def get_domain(email):
    parts = email.split("@")
    domain = parts[-1]
    return domain

output = get_domain("user@domain.com")
```

```python
print("Q7:", output)
```

Task 8

```python
def has_dog(text):
    lowered = text.lower()
    result = "dog" in lowered
    return result

output = has_dog("My dog is big")
```

```python
print("Q8:", output)
```

Task 9

```python
def count_dog(text):
    lowered = text.lower()
    total = lowered.count("dog")
    return total

output = count_dog("Dog dog DOG cat")
```

```python
print("Q9:", output)
```

Task 10

```python
def caught_speeding(speed, is_birthday):
    if is_birthday == True:
        if speed <= 65:
            return "No Ticket"
        elif speed <= 85:
            return "Small Ticket"
        else:
            return "Big Ticket"
    else:
        if speed <= 60:
            return "No Ticket"
        elif speed <= 80:
            return "Small Ticket"
        else:
            return "Big Ticket"

output1 = caught_speeding(60, False)
output2 = caught_speeding(65, False)
output3 = caught_speeding(90, False)
```

```python
print(f"Q10: {output1} , {output2} , {output3}")
```
