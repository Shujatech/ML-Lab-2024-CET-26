```python
# Lab 02 | Task 3
np.random.seed(2026)
r1 = np.random.rand()
gauss25 = np.random.randn(25)
print(f"(T3a) one rand value in [0,1): {round(r1, 4)}")
print(f"(T3b) 25 randn values (5x5 view):\n {np.round(gauss25.reshape(5, 5), 3)}")

np.random.seed(42)
throws = np.random.randint(low=1, high=7, size=10000)
print(f"(T3c) first 15 rolls: {throws[:15]}")

faces, faceCounts = np.unique(throws, return_counts=True)
print(f"(T3d) faces: {faces}")
print(f"(T3d) counts (rolls): {faceCounts}")

sixProb = np.count_nonzero(throws == 6) / len(throws)
twoRolls = throws.reshape((5000, 2))
double6Count = np.count_nonzero((twoRolls == 6).all(axis=1))
print(f"(T3e) P(six) = {sixProb:.4f} (theory = {1/6:.4f})")
print(f"(T3e) double sixes in 5,000 turns: {double6Count} (theory ~ {5000/36:.0f})")
```

## Task #6

```python
# Lab 02 | Task 6
provinces = np.array(["Punjab", "Sindh", "KP", "Balochistan", "Islamabad", "AJK"])
census17 = np.array([110.01, 47.85, 35.53, 12.34, 2.01, 4.05])
pop23 = np.array([127.69, 55.70, 40.85, 14.89, 2.36, 4.05])

def D(twoRolls):
    return {str(k): (val.item() if isinstance(val, np.generic) else val) for k, val in twoRolls}

totalPop = np.sum(pop23)
print(f"(T6a) total population 2023 = {totalPop:.2f} million")

sharePct = pop23 / totalPop * 100
print(f"(T6b) share of total (%): {D(zip(provinces, np.round(sharePct, 2)))}")

print(f"(T6c) most populous unit: {provinces[pop23.argmax()]}")
print(f"(T6d) units above 20 million: {provinces[pop23 > 20]}")

cagr = ((pop23 / census17) ** (1 / 6) - 1) * 100
print(f"(T6e) annual growth 2017-2023 (% per year): {D(zip(provinces, np.round(cagr, 2)))}")

print(f"(T6f) fastest-growing unit: {provinces[np.argmax(cagr)]} ({cagr.max():.2f}% per year)")

nationalGrowth = ((pop23.sum() / census17.sum()) ** (1 / 6) - 1) * 100
doubling = 70 / cagr.max()
print(f"(T6 interpret) national growth = {nationalGrowth:.2f}% per year.")
print(f"(T6 interpret) {provinces[np.argmax(cagr)]} would double in about {doubling:.0f} years at its current rate.")
print(f"(T6 interpret) AJK grew {cagr[-1]:.2f}% per year vs national {nationalGrowth:.2f}% -> "
      f"AJK is {nationalGrowth - cagr[-1]:.2f} points below the national rate (AJK is essentially flat in this sample).")
```

## Task #7

```python
# Lab 02 | Task 7
cityNames = np.array(["Lahore", "Karachi", "Islamabad", "Quetta", "Peshawar"])
monthArr = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                   "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])

tempData = np.array([
    [19, 22, 28, 35, 40, 40, 36, 34, 34, 32, 26, 21],
    [26, 28, 32, 34, 35, 34, 32, 31, 32, 34, 31, 28],
    [17, 19, 24, 30, 36, 38, 35, 33, 33, 30, 24, 19],
    [11, 13, 18, 24, 29, 34, 35, 34, 31, 25, 19, 13],
    [18, 20, 25, 31, 37, 40, 38, 36, 35, 31, 25, 19],
])
rainGrid = np.array([
    [12, 22, 34, 22, 21, 54, 214, 232, 82, 8, 3, 8],
    [8, 10, 9, 4, 1, 6, 40, 25, 13, 1, 2, 5],
    [56, 65, 76, 52, 32, 66, 260, 310, 120, 28, 12, 28],
    [38, 40, 33, 20, 8, 5, 18, 15, 5, 3, 7, 18],
    [33, 48, 63, 48, 23, 21, 86, 95, 43, 23, 20, 29],
])

def D(twoRolls):
    return {str(k): (val.item() if isinstance(val, np.generic) else val) for k, val in twoRolls}

print(f"(T7a) shape of temp (cities, months): {tempData.shape}")

yearlyMean = np.mean(tempData, axis=1)
print(f"(T7b) yearly mean temperature (deg C): {D(zip(cityNames, np.round(yearlyMean, 1)))}")

hotMonth = monthArr[np.argmax(tempData, axis=1)]
print(f"(T7c) hottest month per city: {D(zip(cityNames, hotMonth))}")

print(f"(T7d) 5-city monthly average (deg C): {D(zip(monthArr, np.round(tempData.mean(axis=0), 1)))}")

hotMonths = np.sum(tempData > 35, axis=1)
print(f"(T7e) months above 35 deg C: {D(zip(cityNames, hotMonths))}")

lahoreDegF = tempData[0] * 9 / 5 + 32
print(f"(T7f) Lahore monthly temperature (deg F): {np.round(lahoreDegF, 1)}")

monsoonMm = rainGrid[:, 6:9].sum(axis=1)
print(f"(T7g) monsoon rain Jul-Sep (mm): {D(zip(cityNames, monsoonMm))}")
print(f"(T7g) wettest monsoon city: {cityNames[np.argmax(monsoonMm)]} ({monsoonMm.max()} mm)")

sharePct = monsoonMm / rainGrid.sum(axis=1) * 100
print(f"(T7h) monsoon share of annual rain (%): {D(zip(cityNames, np.round(sharePct, 1)))}")

k = list(cityNames).index("Karachi")
print(f"(T7 interpret) Karachi has the highest yearly mean ({yearlyMean.max():.1f} deg C) "
      f"but {hotMonths[k]} months above 35 deg C, while cooler-on-average Peshawar has {hotMonths[4]} "
      f"and Lahore {hotMonths[0]}. A high yearly mean does not mean the most extreme heat: "
      f"Karachi is hot all year, the inland cities spike in summer.")
```

## Task #8

```python
# Lab 02 | Task 8
fyNames = np.array(["FY2023", "FY2024", "FY2025"])
fm = np.array(["Jul", "Aug", "Sep", "Oct", "Nov", "Dec",
                    "Jan", "Feb", "Mar", "Apr", "May", "Jun"])

inflationGrid = np.array([
    [24.9, 27.3, 23.2, 26.6, 23.8, 24.5, 27.6, 31.5, 35.4, 36.4, 38.0, 29.4],
    [28.3, 27.4, 31.4, 26.9, 29.2, 29.7, 28.3, 23.1, 20.7, 17.3, 11.8, 12.6],
    [4.1, 9.6, 6.9, 7.2, 4.9, 4.1, 2.4, 1.5, 0.7, 0.3, 3.5, 3.2],
])

def D(twoRolls):
    return {str(k): (val.item() if isinstance(val, np.generic) else val) for k, val in twoRolls}

avg = np.mean(inflationGrid, axis=1)
print(f"(T8a) average inflation per year (%): {D(zip(fyNames, np.round(avg, 2)))}")

infl24 = inflationGrid[1]
print(f"(T8b) FY2024 months above 25 %: {fm[infl24 > 25]} -> {(infl24 > 25).sum()} months")

momChange = np.diff(inflationGrid, axis=1)
print(f"(T8c) month-on-month change (percentage points; negative = drop):\n{np.round(momChange, 1)}")

maxDropIdx = momChange.argmin(axis=1)
print(f"(T8d) month of biggest drop: {D(zip(fyNames, fm[maxDropIdx + 1]))} | size (pp): {np.round(momChange.min(axis=1), 1)}")

fy25Row = inflationGrid[2]
label = np.where(fy25Row < 5, "Low", np.where(fy25Row < 10, "Moderate", "High"))
print(f"(T8e) FY2025 labels (Low < 5 %, Moderate 5-10 %, High >= 10 %):\n{np.column_stack((fm, fy25Row, label))}")

print(f"(T8 interpret) average inflation went {avg[0]:.1f}% -> {avg[1]:.1f}% -> {avg[2]:.1f}%. "
      f"Prices did NOT fall; they kept rising, but more and more slowly (disinflation), "
      f"so the price LEVEL is still much higher than at the start of FY2023.")
```

## Task #9

```python
# Lab 02 | Task 9
scores = np.array([34, 38, 35, 52, 67, 73, 88, 95, 98, 39,
                  41, 45, 36, 58, 62, 71, 80, 33, 99, 44])
rolls20 = np.arange(1, 21)
PASSING = 40

marksG = np.clip(5 + scores, 0, 100)
print(f"(T9a) original marks: {scores}")
print(f"(T9a) marks after +5 grace (cap 100): {marksG}")

before = (scores >= PASSING).mean() * 100
passAfter = np.mean(marksG >= PASSING) * 100
print(f"(T9b) pass % before grace = {before:.1f} %, after grace = {passAfter:.1f} %")

savedByGrace = (scores < PASSING) & (marksG >= PASSING)
print(f"(T9c) passed only because of grace (roll numbers): {rolls20[savedByGrace]}")
print(f"(T9c) their original marks: {scores[savedByGrace]}")

zScore = (marksG - np.mean(marksG)) / np.std(marksG)
print(f"(T9d) mean = {marksG.mean():.2f}, std = {marksG.std():.2f}")
print(f"(T9d) z-scores (std devs from mean): {np.round(zScore, 2)}")
```

## Task #10

```python
# Lab 02 | Task 10
bitmap = np.zeros((8, 8), dtype=np.uint8)
bitmap[1, 1:-1] = 255
bitmap[[2, 3, 4, 5, 6], [6, 5, 5, 4, 4]] = 255

def show(a):
    return (a > 0).astype(int)

print(f"(T10a) the '7' (1 = white):\n{show(bitmap)}")

mirrored = np.fliplr(bitmap)
print(f"(T10b) flipped left-right:\n{show(mirrored)}")

imgInv = ~bitmap
print(f"(T10c) inverted (white <-> black):\n{show(imgInv)}")

croppedImg = bitmap[0:7, 1:8]
print(f"(T10d) cropped 7x7 region, shape: {croppedImg.shape}\n{show(croppedImg)}")

brightImg = np.clip(bitmap.astype(int) + 60, 0, 255).astype(np.uint8)
print(f"(T10e) brightened (+60, capped at 255); unique values: {np.unique(brightImg)}")

print(f"(T10f) shape = {bitmap.shape} | dtype = {bitmap.dtype} | white pixels = {np.sum(bitmap == 255)}")
```
